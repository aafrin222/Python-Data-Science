```python
import numpy as np
import time

#python 
l1 = list(range(1000000))
start = time.time()
list_squares = [x ** 2 for x in l1]
end = time.time()
print('python list time :',end-start)

#Using NumpyArry
start = time.time()
arr = np.array(list(range(1000000)))
np_sqyare = arr ** 2 
end = time.time()
print('NumpyArray time :',end-start)
```
