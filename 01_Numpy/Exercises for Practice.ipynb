### Create a 3×3 array filled with random numbers and print its shape.

```python
import numpy as np
arr = np.random.rand(3,3) # float random numbers
# np.random.randint(1, 100, (3, 3)) (integer random numbers)
print(arr)
print("Shape :",arr.shape)
```

### Convert an array of floats [1.1, 2.2, 3.3] into integers.

```python
arr = np.array([1.1,2.2,3.3])
arr_int = arr.astype(int)
print(arr_int)
print(arr_int.dtype)
```

### Use fancy indexing to extract even numbers from [1, 2, 3, 4, 5, 6].

```python
arr = np.array([1,2,3,4,5,6])
even = (arr[arr % 2 == 0])
print(even)
```

### Reshape 1D array of size 9 into 3×3 matrix.

```python
arr = np.arange(1,10)
print("1D array :",arr)
a1 = arr.reshape(3,3)
print(a1)
print("shape of the array is :",a1.shape)
```

### Use boolean masking to filter numbers greater than 50 in an array.

```python
arr = np.array([10, 60, 45, 80, 30, 90])
a2 = arr > 50
print("the greater number of 50 is :",arr[a2])  # print(arr[arr > 50])
```
