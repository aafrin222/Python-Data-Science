## Assignment 2: Side-by-Side and Horizontal Bar Charts

```python
import matplotlib.pyplot as plt
import numpy as np

# Selected 5 years
years = [1990, 1992, 1994, 1996, 1998]

# Runs scored in those years
sachin = [500, 700, 1100, 1500, 1800]
kohli = [0, 0, 500, 800, 1100]
sehwag = [0, 200, 900, 1400, 1600]

# X positions for grouped bars
x = np.arange(len(years))
width = 0.25

# Side-by-side bar chart
plt.bar(x - width, sachin, width=width, label="Sachin", color="skyblue")
plt.bar(x, kohli, width=width, label="Kohli", color="orange")
plt.bar(x + width, sehwag, width=width, label="Sehwag", color="green")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Runs Comparison of Sachin, Kohli, and Sehwag")
plt.xticks(x, years)
plt.legend()
plt.tight_layout()
plt.show()

# Total runs in debut 5 years
players = ["Sachin", "Kohli", "Sehwag"]
total_runs = [sum(sachin), sum(kohli), sum(sehwag)]

for i in range(len(players)):
    plt.text(total_runs[i] + 50, i, str(total_runs[i]), va='center')

# Horizontal bar chart
plt.barh(players, total_runs, color=["skyblue", "orange", "green"])
plt.xlabel("Total Runs in Debut 5 Years")
plt.title("Total Runs in Debut 5 Years")
plt.tight_layout()
plt.show()
```
