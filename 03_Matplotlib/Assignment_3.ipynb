## Assignment3: Mobile OS Market Share using Pie Chart and Horizontal Bar Chart

```python
import matplotlib.pyplot as plt

# Data
labels = ["Android", "iOS", "Others"]
market_share = [72, 27, 1]

# Pie chart
plt.figure(figsize=(6, 6))
plt.pie(
    market_share,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90,
    colors=["skyblue", "lightcoral", "lightgreen"]
)
plt.title("Mobile OS Market Share - Pie Chart")
plt.show()

# Horizontal bar chart
plt.figure(figsize=(7, 4))
plt.barh(labels, market_share, color=["skyblue", "lightcoral", "lightgreen"])
plt.xlabel("Market Share (%)")
plt.title("Mobile OS Market Share - Horizontal Bar Chart")

for i in range(len(labels)):
    plt.text(market_share[i] + 1, i, str(market_share[i]) + "%", va='center')

plt.tight_layout()
plt.show()
```
