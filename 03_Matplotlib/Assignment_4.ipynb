## Weekly Schedule Stack Plot

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

internship = [6, 6, 6, 6, 6, 0, 0]
data_science = [2, 2, 2, 2, 2, 3, 3]
project_work = [1, 1, 1, 1, 1, 2, 2]
deen = [2, 2, 2, 2, 2, 2, 2]
self_care = [1, 1, 1, 1, 1, 2, 2]
sleep = [8, 8, 8, 8, 8, 8, 8]
personal_hobby = [4, 4, 4, 4, 4, 7, 7]

labels = ["Internship", "Data Science", "Project Work", "Deen", "Self Care", "Sleep", "Personal/Hobby"]
colors = ["skyblue", "lightgreen", "gold", "lightcoral", "plum", "gray", "orange"]

plt.figure(figsize=(10, 6))
plt.stackplot(
    days,
    internship,
    data_science,
    project_work,
    deen,
    self_care,
    sleep,
    personal_hobby,
    labels=labels,
    colors=colors,
    alpha=0.8
)

plt.legend(loc="upper left")
plt.title("My Weekly Schedule")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Weekly Schedule Bar Plot

```python
import matplotlib.pyplot as plt
import numpy as np

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

internship = [6, 6, 6, 6, 6, 0, 0]
data_science = [2, 2, 2, 2, 2, 3, 3]
project_work = [1, 1, 1, 1, 1, 2, 2]
deen = [2, 2, 2, 2, 2, 2, 2]
self_care = [1, 1, 1, 1, 1, 2, 2]
sleep = [8, 8, 8, 8, 8, 8, 8]
personal_hobby = [4, 4, 4, 4, 4, 7, 7]

x = np.arange(len(days))
width = 0.1

plt.figure(figsize=(12, 6))

plt.bar(x - 3*width, internship, width=width, label="Internship", color="skyblue")
plt.bar(x - 2*width, data_science, width=width, label="Data Science", color="lightgreen")
plt.bar(x - width, project_work, width=width, label="Project Work", color="gold")
plt.bar(x, deen, width=width, label="Deen", color="lightcoral")
plt.bar(x + width, self_care, width=width, label="Self Care", color="plum")
plt.bar(x + 2*width, sleep, width=width, label="Sleep", color="gray")
plt.bar(x + 3*width, personal_hobby, width=width, label="Personal/Hobby", color="orange")

plt.xlabel("Day")
plt.ylabel("Hours")
plt.title("My Weekly Schedule - Grouped Bar Chart")
plt.xticks(x, days)
plt.legend()
plt.tight_layout()
plt.show()
```

## Observation

- The stack plot shows the overall distribution of activities across the week.
- The grouped bar chart makes it easier to compare exact hours for each activity.
- Sleep hours are much clearer in the grouped bar chart than in the stack plot.
