## Assignment: Histogram of Student Marks

```python
import numpy as np
import matplotlib.pyplot as plt

marks = [35, 42, 50, 55, 60, 61, 65, 67, 70, 72,
         75, 78, 80, 82, 85, 87, 90, 91, 95, 98,
         45, 53, 58, 62, 68, 74, 79, 84, 88, 92]

bins = [30, 40, 50, 60, 70, 80, 90, 100]

plt.hist(marks, bins=bins, edgecolor="black")
plt.title("Distribution of Syudent Marks")
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.grid(True,linewidth=1)
plt.axvline(np.mean(marks), color="red", linestyle="--", linewidth=2, label="Average Marks")
plt.legend()
```
