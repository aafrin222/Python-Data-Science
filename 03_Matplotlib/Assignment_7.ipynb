## Assignment 7: Student Performance Dashboard using Subplots

```python
import matplotlib.pyplot as plt

tests = [1, 2, 3, 4, 5]
marks = [65, 72, 78, 74, 85]

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

# Line plot
axs[0].plot(tests, marks, color="blue", marker="o")
axs[0].set_title("Marks in Line Plot")
axs[0].set_xlabel("Test Number")
axs[0].set_ylabel("Marks")
axs[0].grid(True)

# Bar chart
axs[1].bar(tests, marks, color="orange")
axs[1].set_title("Marks in Bar Chart")
axs[1].set_xlabel("Test Number")
axs[1].set_ylabel("Marks")

# Figure title
fig.suptitle("Student Performance Dashboard", fontsize=14)

fig.tight_layout()
fig.subplots_adjust(top=0.85)
```
