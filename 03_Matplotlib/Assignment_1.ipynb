# Assignment:
Create a plot comparing Kohli, Rohit Sharma, and Sehwag across 10 years of hypothetical runs.
Use:

- Labels
- Legends
- Colors
- Line styles
- One custom style

```python
import matplotlib.pyplot as plt

years = [2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020]

kohli = [800, 950, 1100, 1050, 1200, 1350, 1250, 1400, 1300, 1150]
rohit = [700, 850, 900, 1000, 950, 1100, 1200, 1250, 1400, 1500]
sehwag = [1000, 900, 850, 700, 650, 500, 400, 300, 200, 100]

plt.style.use("ggplot")   # custom style

plt.plot(years, kohli, color="blue", linestyle="-", linewidth=2, label="Virat Kohli")
plt.plot(years, rohit, color="green", linestyle="--", linewidth=2, label="Rohit Sharma")
plt.plot(years, sehwag, color="red", linestyle=":", linewidth=2, label="Virender Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Comparison of Kohli, Rohit Sharma, and Sehwag")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
