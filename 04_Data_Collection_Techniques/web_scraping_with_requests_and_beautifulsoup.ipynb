# Web Scraping using BeautifulSoup

## Objective
The objective of this notebook is to understand the basic process of web scraping using BeautifulSoup in Python.

## Introduction
Web scraping is the process of extracting data from websites.

In Python, BeautifulSoup is commonly used to parse HTML and extract useful information such as:
- titles
- headings
- paragraphs
- links

In this notebook, we will use sample HTML content so that the code runs easily without internet-related errors.

## Step 1: Import Library

```python
from bs4 import BeautifulSoup
```

## Step 2: Create Sample HTML Content
Instead of fetching a live webpage, we will use sample HTML content.

```python
html_content = """
<html>
    <head>
        <title>My Sample Webpage</title>
    </head>
    <body>
        <h1>Welcome to Web Scraping</h1>
        <p>This is the first paragraph.</p>
        <p>This is the second paragraph.</p>
        <a href="https://google.com">Google</a>
        <a href="https://example.com">Example</a>
    </body>
</html>
"""
```

## Step 3: Parse HTML using BeautifulSoup
BeautifulSoup converts raw HTML into a structured format.

```python
soup = BeautifulSoup(html_content, "html.parser")
```

## Step 4: Extract the Page Title
We can extract the title using `soup.title.text`.

```python
print("Page Title:", soup.title.text)
```

## Step 5: Extract the First Heading
The `find()` method returns the first matching tag.

```python
heading = soup.find("h1")
print("First Heading:", heading.text)
```

## Step 6: Extract the First Paragraph

```python
paragraph = soup.find("p")
print("First Paragraph:", paragraph.text)
```

## Step 7: Extract All Paragraphs
The `find_all()` method returns all matching tags.

```python
paragraphs = soup.find_all("p")

for i, p in enumerate(paragraphs, start=1):
    print(f"Paragraph {i}: {p.text}")
```

## Step 8: Extract All Links
We can extract all anchor tags and get their `href` values.

```python
links = soup.find_all("a")

for link in links:
    print(link.get("href"))
```

## Common Methods Used

- `BeautifulSoup(html, "html.parser")` → parses HTML content
- `soup.title` → gets the title tag
- `soup.find("tag")` → finds the first matching tag
- `soup.find_all("tag")` → finds all matching tags
- `.text` → extracts text from a tag
- `.get("attribute")` → extracts attribute values like `href`

## Conclusion
This notebook shows how BeautifulSoup can parse HTML and extract useful data such as titles, headings, paragraphs, and links.

## Key Takeaway
BeautifulSoup helps us read and extract structured information from HTML content in a simple way.
