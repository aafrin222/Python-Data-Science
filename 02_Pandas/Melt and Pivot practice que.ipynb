```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Math": [85, 78, 92],
    "Science": [90, 82, 89],
    "English": [88, 85, 94]
}

df = pd.DataFrame(data)
df
```

# Section 1 — Melt (Wide → Long)
## Question 1

### Convert the DataFrame from wide format to long format using melt().

### Keep Name as the identifier column.

```python
df_long = df.melt(
    id_vars = "Name",
    value_vars = ["Math","Science","English"],
    var_name = "Subject",
    value_name = "Score"
)

df_long
```

# Section 2 — Pivot (Long → Wide)
## Question 2

### Using the melted DataFrame, convert the data back to wide format using pivot().

```python
df_wide = df_long.pivot(
    index="Name",
    columns="Subject",
    values="Score"
)

df_wide
```

# Section 3 — Pivot Table (Handling duplicates)


```python
data2 = {
    "Name": ["Alice", "Alice", "Bob", "Bob"],
    "Subject": ["Math", "Math", "Science", "Science"],
    "Score": [85, 80, 82, 88]
}

df2 = pd.DataFrame(data2)
df2
```

## Question 3

### Use pivot_table() to create a table where:

```python
df_pivot_table = df2.pivot_table(
    index = "Name",
    columns = "Subject",
    values = "Score",
    aggfunc = "mean"
)

df_pivot_table
```
