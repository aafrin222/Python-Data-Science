# Sales Data Analysis using Pandas and NumPy

## Objective
The objective of this project is to analyze retail sales data using Pandas and NumPy.
This project includes data inspection, cleaning, transformation, and analysis.

## Import Libraries

```python
import pandas as pd
import numpy as np
```

## Create dataset

```python
data = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Order_Date": ["2024-01-05", "2024-01-07", "2024-02-10", "2024-02-15", "2024-03-01",
                   "2024-03-05", "2024-03-20", "2024-04-01", "2024-04-07", "2024-04-15"],
    "Customer_Name": ["Ali", "Sara", "John", "Aisha", "Omar", "Zoya", "Ahmed", "Fatima", "Ali", "Sara"],
    "City": ["Ahmedabad", "Surat", "Ahmedabad", "Mumbai", "Surat", "Delhi", "Mumbai", "Ahmedabad", "Surat", np.nan],
    "Category": ["Electronics", "Clothing", "Electronics", "Grocery", "Clothing",
                 "Electronics", "Grocery", "Clothing", "Electronics", "Grocery"],
    "Product": ["Laptop", "T-Shirt", "Mouse", "Rice", "Jeans", "Keyboard", "Oil", "Kurti", "Monitor", "Sugar"],
    "Quantity": [1, 2, 3, 5, 2, 1, 4, 2, 1, 3],
    "Price": [55000, 800, 500, 60, 1200, 1500, 140, 900, 12000, 45],
    "Payment_Mode": ["UPI", "Cash", "Card", "UPI", "Cash", "Card", "UPI", "Cash", "Card", "UPI"]
}

df = pd.DataFrame(data)
df
```

## Create Total Sales Column

```python
df["Total_Sales"] = df["Quantity"] * df["Price"]
df
```

## Checking Missing Values

```python
df.isnull().sum()
```

## Handling Missing Values

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
df
```

```python
df.isnull().sum()
```

## Converting Order_Date to Datetime

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df.info()
```

## Feature Engineering

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Day_Name"] = df["Order_Date"].dt.day_name()

df
```

## Final Insights

- Electronics category generated the highest revenue among all categories.
- Ahmedabad, Surat, and Mumbai contributed significantly to total sales.
- UPI and Cash were commonly used payment modes in the dataset.
- Some products such as Laptop and Monitor contributed much more revenue than low-price grocery products.
- Monthly sales analysis helps identify which months performed better.
