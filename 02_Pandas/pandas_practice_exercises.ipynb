## Q23

## Create new column:

#### Age_Group

#### Rule:

##### Age >= 25 → Adult
##### Age < 25 → Young

```python
def isyoung(x):
    return "Adult" if x>= 25 else "Young"

df["Age Group"] =  df["Age"].apply(isyoung)
df
```

## Q24

## Create ranking based on Score
