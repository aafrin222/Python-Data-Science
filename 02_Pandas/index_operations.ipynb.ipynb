```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Zain"],
    "Age": [22, 24, 23],
    "City": ["Surat", "Ahmedabad", "Baroda"]
})

print("Original DataFrame")
print(df)

df1 = df.set_index("Name")
print("\nAfter set_index")
print(df1)

df2 = df1.reset_index()
print("\nAfter reset_index")
print(df2)
```
